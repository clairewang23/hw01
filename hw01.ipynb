```julia
# function to remove mean from a vector
function remove_mean(vect)
    # fucntion to compute the mean
    function compute_mean(vect)
        element_sum = 0 # initialize sum
        # compute mean and return
        for v in vect
            element_sum += v
        end
        return element_sum / length(vect)
    end

    m = compute_mean(vect) # compute mean
    # return demeaned vector
    return vect .- m
end

random_vect = rand(1_000)
@show remove_mean(random_vect)
```

Make sure you comment your code appropriately to make it clear what is
going on and why.

```julia
function yuk_discharge(x1_f, x2_f)
    # function to compute the mass of YUK discharged into Pristine Brook (kg/d)
    # convert input fractions to discharge values
    x_total = 100 # total wastewater produced by plant, m^3/d
    x1 = x_total*x1_f
    x2 = x_total*x2_f
    return 0.2*x1 + 0.005*(x2^2) + (x_total - x1 - x2) #kg/d
end

function calc_cost(x1_f, x2_f)
    # function to calculate the treatment calc_cost ($/d)
    # convert input fractions to discharge values
    x_total = 100 # total wastewater produced by plant, m^3/d
    x1 = x_total*x1_f
    x2 = x_total*x2_f
    return (x1^2)/20 + 1.5*x2
end
```

Use your function to experiment with 1,000 different combinations of
wastewater discharge and treatment. You can do this with either a grid
search or by sampling from a [Dirichlet
distribution](https://en.wikipedia.org/wiki/Dirichlet_distribution) (a
$\text{Dirichlet}(1, n)$ distribution will generate uniformly-weighted
$n$-dimensional vectors whose components add up to 1; see the
[`Distributions.jl`
documentation](https://juliastats.org/Distributions.jl/stable/starting/)
for how to sample from probability distributions in Julia).

Plot the results of these experiments by making a scatterplot of YUK
discharge into the lake vs. treatment cost. Make sure to label your axes
and include a legend, as well as a caption clearly describing the plot.

```julia
# define Dirichlet distribution of 3 values that add to 1
d = Dirichlet(3,1)

# generate 1000 combinations using Dirichlet distribution. 
n = 1000 # number of trials
x = rand(d, n)

# initialize arrays to store discharge and treatment cost values
discharge = ones(n)
cost = ones(n)

# loop to calculate discharge and cost values for random x1 and x2 values
for i=1:n
    # get values from x
    x1 = x[1,i]
    x2 = x[2,i]

    # calculate discharge and cost
    discharge[i] = yuk_discharge(x1, x2)
    cost[i] = calc_cost(x1, x2)
end

# plot results
#plot(discharge, cost, seriestype = :scatter, xlabel="YUK discharge into creek (kg/d)", ylabel="Treatment cost (\$/d)", label="Results from discharge combinations")
plot(cost, discharge, seriestype = :scatter, ylabel="YUK discharge into creek (kg/d)", xlabel="Treatment cost (\$/d)", label="Results from discharge combinations")
hline!([20], linestyle = :dot, linewidth = 2, label = "YUK effluent standard")
```
